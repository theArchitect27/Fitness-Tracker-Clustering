```python
# Clustering Fitness Tracker Data using KMeans
# Author: Tlhologelo Mmako

import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python


active_num = np.random.randint(1, 11) * 100
moderate_num = np.random.randint(1, 11) * 100
least_active_num= np.random.randint(1, 11) * 100

# Cluster 1 (Active)
steps_active = np.random.uniform(8000, 12000, active_num)
sleep_active = np.random.uniform(7, 9, active_num)

# Cluster 2 (Moderate)
steps_moderate = np.random.uniform(5000, 8000, moderate_num)
sleep_moderate = np.random.uniform(6, 7.5, moderate_num)

# Cluster 3 (Least active)
steps_least = np.random.uniform(2000, 5000, least_active_num)
sleep_least = np.random.uniform(5, 6.5, least_active_num)
```

```python


steps_per_day = np.concatenate([steps_active, steps_moderate, steps_least])
hours_slept = np.concatenate([sleep_active, sleep_moderate, sleep_least])

data = np.column_stack((steps_per_day, hours_slept))
```

```python

kmeans = KMeans(n_clusters=3)
kmeans.fit(data)

labels = kmeans.labels_
centroids = kmeans.cluster_centers_
```

```python

plt.scatter(steps_per_day, hours_slept, c=labels)
plt.scatter(centroids[:, 0], centroids[:, 1], marker='X', s=200)
plt.xlabel("Steps per Day")
plt.ylabel("Hours Slept per Night")
plt.title("KMeans Clustering of Fitness Tracker Data")
plt.show()
```

```python

for i in range(3):
    cluster_points = data[labels == i]

    avg_steps = np.mean(cluster_points[:, 0])
    avg_sleep = np.mean(cluster_points[:, 1])

    print(f"Cluster {i+1}")
    print(f"Average Steps per Day: {avg_steps:.2f}")
    print(f"Average Hours Slept per Night: {avg_sleep:.2f}")
    print()
```
